# 1. Create a Python program that simulates a simple AI agent that can learn to play tic-tac-toe. 
The program must Define the game board, Check if a player has won, Check if the game is a tie, Main game loop, Call the main game loop,                                                                                                                                                                               

```python
#Defining the game board_ We represent the board as a list of 9 elements
# Create an empty Tic-Tac-Toe board
board = [" " for _ in range(9)]
```

 Index reference for the player

0 | 1 | 2
---------
3 | 4 | 5
---------
6 | 7 | 8


```python
#Display the board

def print_board(board):
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()
```

```python
# Check all the winning combinations to know if the player has won
def check_winner(board, player):
    win_combinations = [
        (0,1,2), (3,4,5), (6,7,8),  # rows
        (0,3,6), (1,4,7), (2,5,8),  # columns
        (0,4,8), (2,4,6)            # diagonals
    ]
    
    for combo in win_combinations:
        if board[combo[0]] == board[combo[1]] == board[combo[2]] == player:
            return True
    return False
```

```python
# Check if there is a tie i.e. when the board is full and no winner exists
def check_tie(board):
    return " " not in board
```

Simple AI Agent (Logic);

It tries to win

It tries to block the player

It chooses a random move otherwise

```python
import random

def ai_move(board):
    # Try to win
    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            if check_winner(board, "O"):
                return
            board[i] = " "

    # Try to block player
    for i in range(9):
        if board[i] == " ":
            board[i] = "X"
            if check_winner(board, "X"):
                board[i] = "O"
                return
            board[i] = " "

    # Random move
    empty_positions = [i for i in range(9) if board[i] == " "]
    board[random.choice(empty_positions)] = "O"
```

Main Game Loop

This controls turns, checks game state, and prints results

```python
def play_game():
    global board
    board = [" " for _ in range(9)]

    print("Welcome to Tic-Tac-Toe!")
    print("You are X, AI is O")
    print_board(board)

    while True:
        # Player move
        move = int(input("Choose a position (0-8): "))
        if board[move] != " ":
            print("Position already taken. Try again.")
            continue
        
        board[move] = "X"
        print_board(board)

        if check_winner(board, "X"):
            print("You win! 🎉")
            break

        if check_tie(board):
            print("It's a tie! 🤝")
            break

        # AI move
        print("AI is making a move...")
        ai_move(board)
        print_board(board)

        if check_winner(board, "O"):
            print("AI wins! 🤖")
            break

        if check_tie(board):
            print("It's a tie! 🤝")
            break
```

Call the Main Game Loop (To Play the Game)

```python
play_game()
```
