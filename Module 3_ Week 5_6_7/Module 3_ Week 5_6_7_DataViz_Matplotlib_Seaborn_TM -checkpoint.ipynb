 ### Tipping Behaviour Analysis

### Project Objective

- Analyze tipping behavior in a restaurant dataset to uncover how factors such as total bill amount, customer demographics (gender) and dining context (day and time) influence tip size.
- Use data visualization techniques to identify trends, distributions and relationships that can inform customer behavior insights.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
# Load dataset
tips = sns.load_dataset("tips")



```

```python
#quick view of the dataset
tips.info()
tips.describe()
```

```python
tips.head()
```

```python
# check for missing values
tips.isnull().sum()
```

### Trend Analysis

```python

# Line Plot (Average total bill by day)

avg_bill = tips.groupby("day", observed=False)["total_bill"].mean()

plt.figure(figsize=(6,4))
avg_bill.plot(marker='o', linestyle='-', color='green')
plt.title("Average Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Average Total Bill ($)")
plt.grid(True)
plt.show()

```

### Insight 

- Saturdays and Sundays shows higher average spending as compared to Thursdays and Fridays

### Relationship Analysis

```python

# Scatter Plot (total bill vs tip)

plt.figure(figsize=(6,4))
plt.scatter(tips["total_bill"], tips["tip"], color='green', alpha=0.6)
plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.show()
```

```python
sns.scatterplot(x="total_bill", y="tip", hue="time", data=tips)
plt.title("Total Bill vs Tip by Time")
plt.show()
```

 ## Insights

- Most of the tips received were below $5
- More tips were received during dinner time
- Tips increase with total bill, indicating a positive correlation
- Some tips are outliers, suggesting occasional very generous tippings


### Distribution Analysis

```python

# Histogram (total bill)

plt.figure(figsize=(6,4))
sns.histplot(tips["total_bill"], kde=True, bins=20, color='green')
plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill ($)")
plt.ylabel("Frequency")
plt.show()

```

## Insights

- The distribution of total bills is slightly right-skewed i.e. there are more lower bills
- The highest frequency of total bills falls within the range of 10 - 20 dollar

### Comparative Analysis

```python

# Box plot (tip by day and gender)

# by day
plt.figure(figsize=(6,4))
sns.boxplot(x="day", y="tip", data=tips, color='green')
plt.title("Tip Distribution by Day")
plt.show()

# by gender
sns.boxplot(x="sex", y="tip", data=tips, color='green')
plt.title("Tip Distribution by Gender")
plt.show()
```

## Insights

- Sunday shows a higher tip distribution as compared to Thursday, Friday and Saturday
- Male customers tend to tip slightly more than the female customers
