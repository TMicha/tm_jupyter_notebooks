# Create a Python program that simulates a game of Rock, Paper, Scissors between the user and the computer. 

The program must 
Define the possible choices, 
Get the user's choice, 
Generate a random choice for the computer, 
Determine the winner 
based on the rules of Rock, Paper, Scissors.


```python
#Storing the valid options in a list
#To define possible choices
choices = ["rock", "paper", "scissors"]
```

```python
#Prompt the user for input...it's being converted to lower case to avoid case sensitivity issues
user_choice = input("Enter your choice (rock, paper, scissors): ").lower()

# Validate user entry
if user_choice not in choices:
    print("Invalid choice. Please choose rock, paper or scissors.")
```

```python
# To simulate the computer's move, generate random choices

import random

computer_choice = random.choice(choices)
```

```python
# Compare both choices using conditional logic that follows the game rules to
# determine the winner
print(f"\nYou chose: {user_choice}")
print(f"Computer chose: {computer_choice}")

if user_choice == computer_choice:
    print("It's a tie!")

elif (
    (user_choice == "rock" and computer_choice == "scissors") or
    (user_choice == "scissors" and computer_choice == "paper") or
    (user_choice == "paper" and computer_choice == "rock")
):
    print("You win!")

else:
    print("Computer wins!")

```
